```python
# ==============================
#     FORM DATA MAHASISWA
# ==============================

def form_mahasiswa():
    clear_frame()

    judul = tk.Label(root, text="FORM DATA MAHASISWA", font=("Arial", 16, "bold"), fg="#1A237E")
    judul.pack(pady=20)

    
    frame_atas = tk.Frame(root)
    frame_atas.pack(pady=10, fill=tk.X, padx=20)

    
    frame_input = tk.LabelFrame(frame_atas, text="Input Data Mahasiswa", padx=10, pady=10)
    frame_input.pack(side="left", fill=tk.BOTH, expand=True, padx=(0,10))

    frame_search = tk.LabelFrame(frame_atas, text="Pencarian", padx=10, pady=10)
    frame_search.pack(side=tk.RIGHT, fill=tk.BOTH, padx=(10,0))


    tk.Label(frame_input, text="NIM").grid(row=0, column=0, sticky="w", padx=5, pady=3)
    entry_nim = tk.Entry(frame_input, width=35)
    entry_nim.grid(row=0, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Nama Mahasiswa").grid(row=1, column=0, sticky="w", padx=5, pady=3)
    entry_nama = tk.Entry(frame_input, width=35)
    entry_nama.grid(row=1, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Program Studi").grid(row=2, column=0, sticky="w", padx=5, pady=3)
    entry_prodi = tk.Entry(frame_input, width=35)
    entry_prodi.grid(row=2, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Jenis Kelamin").grid(row=3, column=0, sticky="w", padx=5, pady=3)
    combo_jk = ttk.Combobox(frame_input, values=["L", "P"], width=32, state="readonly")
    combo_jk.grid(row=3, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Tempat Lahir").grid(row=4, column=0, sticky="w", padx=5, pady=3)
    entry_tempat = tk.Entry(frame_input, width=35)
    entry_tempat.grid(row=4, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Tanggal Lahir").grid(row=5, column=0, sticky="w", padx=5, pady=3)
    entry_tanggal = tk.Entry(frame_input, width=35)
    entry_tanggal.grid(row=5, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Alamat").grid(row=6, column=0, sticky="w", padx=5, pady=3)
    entry_alamat = tk.Entry(frame_input, width=35)
    entry_alamat.grid(row=6, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Desa/Kelurahan").grid(row=7, column=0, sticky="w", padx=5, pady=3)
    entry_desa = tk.Entry(frame_input, width=35)
    entry_desa.grid(row=7, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Kecamatan").grid(row=8, column=0, sticky="w", padx=5, pady=3)
    entry_kecamatan = tk.Entry(frame_input, width=35)
    entry_kecamatan.grid(row=8, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Kabupaten/Kota").grid(row=9, column=0, sticky="w", padx=5, pady=3)
    entry_kabupaten = tk.Entry(frame_input, width=35)
    entry_kabupaten.grid(row=9, column=1, padx=5, pady=3)

    tk.Label(frame_input, text="Provinsi").grid(row=10, column=0, sticky="w", padx=5, pady=3)
    entry_provinsi = tk.Entry(frame_input, width=35)
    entry_provinsi.grid(row=10, column=1, padx=5, pady=3)


    tk.Label(frame_search, text="Cari Nama/NIM").pack(anchor="w", pady=2)
    entry_search = tk.Entry(frame_search, width=25)
    entry_search.pack(pady=5)

# ==============================
#       FUNGSI SEARCH
# ==============================
    
    def search_data():
        kata_kunci = entry_search.get()
        show_data(keyword=kata_kunci)

    tk.Button(frame_search, text="Search", command=search_data, bg="lightgray", width=10).pack(pady=5)
    tk.Button(frame_search, text="Refresh", command=lambda: [entry_search.delete(0, tk.END), show_data()], bg="white", width=10).pack(pady=5) 

# ==============================
#       FUNGSI BERSIHKAN
# ==============================

    def bersihkan():
        entry_nim.config(state="normal")
        entry_nim.delete(0, tk.END)
        entry_nama.delete(0, tk.END)
        entry_prodi.delete(0, tk.END)
        combo_jk.set("")
        entry_tempat.delete(0, tk.END)
        entry_tanggal.delete(0, tk.END)
        entry_alamat.delete(0, tk.END)
        entry_desa.delete(0, tk.END)
        entry_kecamatan.delete(0, tk.END)
        entry_kabupaten.delete(0, tk.END)
        entry_provinsi.delete(0, tk.END)

# ==============================
#       FUNGSI SIMPAN
# ==============================

    def simpan():
        if entry_nim.get() == "" or entry_nama.get() == "" :
            messagebox.showwarning("Peringatan", "NIM dan Nama tidak boleh kosong")
            return  
        
        try:
            conn = get_connection()
            cursor = conn.cursor()

            cursor.execute("SELECT * FROM tbmahasiswa WHERE nim=%s", (entry_nim.get(),))
            if cursor.fetchone():
                messagebox.showwarning("Peringatan", "NIM sudah terdaftar")
                conn.close()
                return
            sql = """INSERT INTO tbmahasiswa (nim, nama_mahasiswa, prodi, jenis_kelamin, tempat_lahir, tanggal_lahir, alamat, desa_kelurahan, kecamatan, kabupaten_kota, provinsi)
                     VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)"""
        
            data = (entry_nim.get(), entry_nama.get(), entry_prodi.get(), combo_jk.get(), entry_tempat.get(), entry_tanggal.get(), entry_alamat.get(), entry_desa.get(), entry_kecamatan.get(), entry_kabupaten.get(), entry_provinsi.get())
            cursor.execute(sql, data)
            conn.commit()
            messagebox.showinfo("Sukses", "Data mahasiswa berhasil disimpan")
            show_data()
            bersihkan()
        except Exception as e:
            messagebox.showerror("Error", f"Gagal Menyimpan data: {e}")
        finally:
            conn.close()
      
# ==============================
#       FUNGSI UPDATE
# ==============================

    def update_data():
        if entry_nim.get() == "" :
            messagebox.showwarning("Peringatan", "Pilih data yang ingin diupdate")
            return
        try:
            conn = get_connection()
            cursor = conn.cursor()
            sql = """UPDATE tbmahasiswa SET nama_mahasiswa=%s, prodi=%s, jenis_kelamin=%s, tempat_lahir=%s, tanggal_lahir=%s, alamat=%s, desa_kelurahan=%s, kecamatan=%s, kabupaten_kota=%s, provinsi=%s WHERE nim=%s"""
            data = (entry_nama.get(), entry_prodi.get(), combo_jk.get(), entry_tempat.get(), entry_tanggal.get(), entry_alamat.get(), entry_desa.get(), entry_kecamatan.get(), entry_kabupaten.get(), entry_provinsi.get(), entry_nim.get())
            cursor.execute(sql, data)
            conn.commit()
            messagebox.showinfo("Sukses", "Data mahasiswa berhasil diupdate")
            show_data()
            bersihkan()
        except Exception as e:
            messagebox.showerror("Error", f"Gagal Update data: {e}")
        finally:
            conn.close()    

       
# ==============================
#       FUNGSI DELETE DATA
# ==============================
   
    def delete_data():
        selected_item = tree.selection()
        if not selected_item:
            messagebox.showwarning("Peringatan", "Pilih data yang ingin dihapus")
            return
        
        data = tree.item(selected_item)["values"]
        nim = data[0]
        nama = data[1]
        if messagebox.askyesno("Konfirmasi", f"Apakah Anda yakin ingin menghapus data dengan NIM {nama} {nim}?"):
            try:
                conn = get_connection()
                cursor = conn.cursor()
                cursor.execute("DELETE FROM tbmahasiswa WHERE nim=%s", (nim,))
                conn.commit()
                messagebox.showinfo("Sukses", "Data mahasiswa berhasil dihapus")
                show_data()
                bersihkan()
            except Exception as e:
                messagebox.showerror("Error", f"Gagal Menghapus data: {e}")
            finally:
                conn.close()
     
# ==============================
#       FUNGSI SHOW DATA
# ==============================
  
    def show_data(keyword=None):
        for item in tree.get_children():
            tree.delete(item)
        try:    
            conn = get_connection()
            cursor = conn.cursor()
            if keyword:
                query = """SELECT nim, nama_mahasiswa, prodi, jenis_kelamin, tempat_lahir, tanggal_lahir, alamat, desa_kelurahan, kecamatan, kabupaten_kota, provinsi 
                           FROM tbmahasiswa WHERE nama_mahasiswa LIKE %s OR nim LIKE %s"""
                cursor.execute(query, (f"%{keyword}%", f"%{keyword}%"))
            else:
                cursor.execute("SELECT nim, nama_mahasiswa, prodi, jenis_kelamin, tempat_lahir, tanggal_lahir, alamat, desa_kelurahan, kecamatan, kabupaten_kota, provinsi FROM tbmahasiswa")
         
            rows = cursor.fetchall() 
            for count, row in enumerate(rows):
                if count % 2 == 0:
                    tree.insert("","end", values=row, tags=("evenrow",)) 
                else:
                    tree.insert("","end", values=row, tags=("oddrow",))
        except Exception as e:
            messagebox.showerror("Error", f"Gagal Menampilkan data: {e}")
        finally:
            conn.close()
        

# ==============================
#       FUNGSI PILIH DATA
# ==============================

    def pilih_data(event):
        selected = tree.focus()
        data = tree.item(selected, "values")
        if data:
            bersihkan()

            entry_nim.insert(0, data[0])
            entry_nim.config(state="disabled")
            entry_nama.insert(0, data[1])
            entry_prodi.insert(0, data[2])
            combo_jk.set(data[3])
            entry_tempat.insert(0, data[4])
            entry_tanggal.insert(0, data[5])
            entry_alamat.insert(0, data[6]) 
            entry_desa.insert(0, data[7])
            entry_kecamatan.insert(0, data[8])
            entry_kabupaten.insert(0, data[9])
            entry_provinsi.insert(0, data[10])



# ==============================
#           TOMBOL
# ==============================
   
    frame_buttons = tk.Frame(frame_input)
    frame_buttons.grid(row=11, column=1, pady=15, sticky="e")

    tk.Button(frame_buttons, text="SAVE", command=simpan, bg="#4CAF50", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_buttons, text="UPDATE", command=update_data, bg="#2196F3", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_buttons, text="DELETE", command=delete_data, bg="#f44336", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_buttons, text="CLEAR", command=bersihkan, bg="#FF9800", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_buttons, text="SHOW DATA", command=show_data, bg="#9C27B0", fg="white", width=10, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_buttons, text="HOME", command=halaman_utama, bg="#9E9E9E", fg="white", width=8).pack(side=tk.LEFT, padx=3)

# ==============================
#           TREEVIEW
# ==============================
   
    frame_table = tk.Frame(root)
    frame_table.pack(pady=10, fill=tk.BOTH, expand=True, padx=20)

    scroll_y = tk.Scrollbar(frame_table, orient=tk.VERTICAL)
    scroll_y.pack(side=tk.RIGHT, fill=tk.Y)

    scroll_x = tk.Scrollbar(frame_table, orient=tk.HORIZONTAL)
    scroll_x.pack(side=tk.BOTTOM, fill=tk.X)

    tree = ttk.Treeview(frame_table, columns=("nim", "nama_mahasiswa", "prodi", "jenis_kelamin","tempat_lahir","tanggal_lahir", "alamat", "desa_kelurahan", "kecamatan", "kabupaten_kota", "provinsi"), show="headings", height=8, yscrollcommand=scroll_y.set, xscrollcommand=scroll_x.set)

    scroll_y.config(command=tree.yview)
    scroll_x.config(command=tree.xview)


    tree.heading("nim", text="NIM")
    tree.heading("nama_mahasiswa", text="Nama Mahasiswa")
    tree.heading("prodi", text="Program Studi")
    tree.heading("jenis_kelamin", text="L/P")
    tree.heading("tempat_lahir", text="Tempat Lahir")
    tree.heading("tanggal_lahir", text="Tanggal Lahir")
    tree.heading("alamat", text="Alamat")
    tree.heading("desa_kelurahan", text="Desa/Kel")
    tree.heading("kecamatan", text="Kecamatan")
    tree.heading("kabupaten_kota", text="Kab/Kota")
    tree.heading("provinsi", text="Provinsi")

    
    tree.column("nim", width=80, anchor="center")
    tree.column("nama_mahasiswa", width=160, anchor="w")
    tree.column("prodi", width=100, anchor="w")
    tree.column("jenis_kelamin", width=40, anchor="center")
    tree.column("tempat_lahir", width=100, anchor="w")
    tree.column("tanggal_lahir", width=90, anchor="center")
    tree.column("alamat", width=150, anchor="w")
    tree.column("desa_kelurahan", width=100, anchor="w")
    tree.column("kecamatan", width=100, anchor="w")
    tree.column("kabupaten_kota", width=100, anchor="w")
    tree.column("provinsi", width=100, anchor="w")

    tree.pack(fill=tk.BOTH, expand=True)

    tree.bind("<ButtonRelease-1>", pilih_data)
    tree.tag_configure("oddrow", background="white")
    tree.tag_configure("evenrow", background="lightblue")





```

```python
# ==============================
#        FUNGSI KONVERSI
# ==============================

def konversi_nilai(nilai):
    nilai = float(nilai)

    if nilai >= 85:
        return "A", 4.00
    elif nilai >= 75:
        return "B", 3.00
    elif nilai >= 65:
        return "C", 2.00
    elif nilai >= 50:
        return "D", 1.00
    else:
        return "E", 0.00
    
```

```python
# ==============================
#      FORM SIMPAN NILAI
# ==============================

def form_nilai():
    clear_frame()

    judul = tk.Label(root, text="FORM SIMPAN NILAI", font=("Arial", 16, "bold"), fg="#1A237E")
    judul.pack(pady=20)

    frame_atas = tk.Frame(root)
    frame_atas.pack(pady=10, fill=tk.X, padx=20)

    frame_input = tk.LabelFrame(frame_atas, text=" Input Data Nilai ", padx=10, pady=10)
    frame_input.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=(0, 10))

    frame_search = tk.LabelFrame(frame_atas, text=" Pencarian Nilai ", padx=10, pady=10)
    frame_search.pack(side=tk.RIGHT, fill=tk.BOTH, padx=(10, 0))


# ==============================
#     FUNGSI AMBIL DATA KRS
# ==============================

    def get_list_krs():
        conn = None
        try: 
            conn = get_connection()
            cursor = conn.cursor()
            
            query ="""SELECT k.id_krs, m.nama_mahasiswa, mk.nama_matakuliah
                      FROM tbkrs k
                      JOIN tbmahasiswa m ON k.nim = m.nim
                      JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk"""
            cursor.execute(query)
            return [f"{row[0]} - {row[1]} - {row[2]}" for row in cursor.fetchall()]
        except: 
            return []
        finally:
            if conn: conn.close()

# ==============================
#            INPUT
# ==============================

    tk.Label(frame_input, text="Pilih KRS").grid(row=0, column=0, sticky="w", padx=5, pady=5)
    combo_krs = ttk.Combobox(frame_input, values=get_list_krs(), width=45, state="readonly")
    combo_krs.grid(row=0, column=1, padx=5, pady=5, sticky="w")

    tk.Label(frame_input,text="Nilai Angka").grid(row=1, column=0, sticky="w", padx=5, pady=5)
    entry_nilai = tk.Entry(frame_input, width=20)
    entry_nilai.grid(row=1, column=1, padx=5, pady=5, sticky="w")


# ==============================
#            SEARCH
# ==============================

    tk.Label(frame_search, text="Cari NIM/Nama Mahasiswa:").pack(anchor="w", pady=2)
    entry_search = tk.Entry(frame_search, width=25)
    entry_search.pack(pady=5)

    frame_btn_search = tk.Frame(frame_search)
    frame_btn_search.pack(pady=5, anchor="w")

    tk.Button(frame_search, text="Search", command=lambda: search_nilai(), bg="lightgray", width=10).pack(pady=5)
    tk.Button(frame_search, text="Refresh", command=lambda: [entry_search.delete(0, tk.END), show_data()], bg="white", width=10).pack(pady=5)

# ==============================
#    FUNGSI SIMPAN DATA NILAI
# ==============================

    def simpan_nilai():
        if not combo_krs.get() or not entry_nilai.get():
            messagebox.showwarning("Peringatan", "Semua data wajib diisi!")
            return
        
        conn = None
        try:
            id_krs = combo_krs.get().split(" - ")[0]
            nilai_angka = float(entry_nilai.get())
            huruf, bobot = konversi_nilai(nilai_angka)

            conn = get_connection()
            cursor = conn.cursor()

            query_sks = """ SELECT mk.sks FROM tbkrs k
                            JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk
                            WHERE k.id_krs = %s """
            cursor.execute(query_sks, (id_krs,))
            result = cursor.fetchone()
            sks = result[0]
            mutu = sks * bobot

            cursor.execute("SELECT * FROM tbnilai WHERE id_krs = %s", (id_krs,))
            if cursor.fetchone():
                messagebox.showerror("Error", "Data nilai untuk KRS ini sudah ada!")
                return
            
            sql = """INSERT INTO tbnilai (id_krs, nilai_angka, nilai_huruf, bobot, mutu)
                     VALUES (%s, %s, %s, %s, %s)"""
            data = (id_krs, nilai_angka, huruf, bobot, mutu)
            cursor.execute(sql, data)
            conn.commit()
            messagebox.showinfo("Sukses", "Data nilai berhasil disimpan.")
            show_data()

        except Exception as e:
            messagebox.showerror("Error", f"Gagal menyimpan data: {e}")
        finally:
            if conn: conn.close()

# ==============================
#    FUNGSI SHOW DATA NILAI
# ==============================

    mapping_id_nilai = {}

    def show_data():
        nonlocal mapping_id_nilai
        mapping_id_nilai.clear()
        
        for item in tree.get_children():
            tree.delete(item)

        conn = None
        try:
            conn = get_connection()
            cursor = conn.cursor()
            query = """SELECT id_nilai, m.nama_mahasiswa, mk.nama_matakuliah, n.nilai_angka, n.nilai_huruf, n.bobot, n.mutu FROM tbnilai n
                       JOIN tbkrs k ON n.id_krs = k.id_krs
                       JOIN tbmahasiswa m ON k.nim = m.nim
                       JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk"""
            
            cursor.execute(query)
            rows = cursor.fetchall()

            for count, row in enumerate(rows):
                nomor_urut = count + 1
                mapping_id_nilai[nomor_urut] = row[0]
                
                data_tampil = (nomor_urut,) + row[1:]

                if count % 2 == 0:
                    tree.insert("", "end", values=data_tampil, tags=("evenrow",))
                else:
                    tree.insert("", "end", values=data_tampil, tags=("oddrow",))
        except Exception as e:
            messagebox.showerror("Error", f"Gagal menampilkan data nilai: {e}")
        finally:
            if conn: conn.close()

# ==============================
#      FUNGSI DELETE NILAI
# ==============================

    def delete_nilai():
        selected_item = tree.selection()

        if not selected_item:
            messagebox.showwarning("Peringatan", "Pilih data nilai yang ingin dihapus!")
            return

        data = tree.item(selected_item)["values"]

        nomor_tabel = data[0]
        id_nilai = mapping_id_nilai.get(nomor_tabel)
        nama_mahasiswa = data[1]

        konfirmasi = messagebox.askyesno(
            "Konfirmasi",
            f"Yakin ingin menghapus nilai milik {nama_mahasiswa}?"
        )

        if konfirmasi:
            conn = None
            try:
                conn = get_connection()
                cursor = conn.cursor()

                cursor.execute(
                    "DELETE FROM tbnilai WHERE id_nilai = %s",
                    (id_nilai,)
                )

                conn.commit()
                messagebox.showinfo("Sukses", "Data nilai berhasil dihapus.")
                show_data()

            except Exception as e:
                messagebox.showerror("Error", f"Gagal menghapus data: {e}")
            finally:
                if conn: conn.close()

# ==============================
#      FUNGSI SEARCH NILAI
# ==============================

    def search_nilai():
        nonlocal mapping_id_nilai
        mapping_id_nilai.clear()
        keyword = entry_search.get()

        for item in tree.get_children():
            tree.delete(item)

        conn = None
        try:
            conn = get_connection()
            cursor = conn.cursor()

            query = """
                SELECT n.id_nilai,
                       m.nama_mahasiswa,
                       mk.nama_matakuliah,
                       n.nilai_angka,
                       n.nilai_huruf,
                       n.bobot,
                       n.mutu
                FROM tbnilai n
                JOIN tbkrs k ON n.id_krs = k.id_krs
                JOIN tbmahasiswa m ON k.nim = m.nim
                JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk
                WHERE m.nama_mahasiswa LIKE %s
                OR mk.nama_matakuliah LIKE %s
            """

            cari = (
                "%" + keyword + "%",
                "%" + keyword + "%"
            )

            cursor.execute(query, cari)
            rows = cursor.fetchall()

            for count, row in enumerate(rows):
                nomor_urut = count + 1
                mapping_id_nilai[nomor_urut] = row[0]
                
                data_tampil = (nomor_urut,) + row[1:]

                if count % 2 == 0:
                    tree.insert("", "end", values=data_tampil, tags=("evenrow",))
                else:
                    tree.insert("", "end", values=data_tampil, tags=("oddrow",))

        except Exception as e:
            messagebox.showerror("Error", f"Gagal mencari data: {e}")
        finally:
            if conn: conn.close()

# ==============================
#        FUNGSI PILIH DATA
# ==============================

    selected_id_nilai = None

    def pilih_data(event):
        nonlocal selected_id_nilai
        selected = tree.focus()
        data = tree.item(selected, "values")

        if data:
            nomor_tabel = data[0]
            selected_id_nilai = mapping_id_nilai.get(nomor_tabel)
            entry_nilai.delete(0, tk.END)
            entry_nilai.insert(0, data[3])

# ==============================
#      FUNGSI UPDATE NILAI
# ==============================

    def update_nilai():
        nonlocal selected_id_nilai

        if not selected_id_nilai:
            messagebox.showwarning("Peringatan", "Pilih data nilai yang ingin diupdate!")
            return

        conn = None
        try:
            nilai_angka = float(entry_nilai.get())
            huruf, bobot = konversi_nilai(nilai_angka)

            conn = get_connection()
            cursor = conn.cursor()

            query_sks = """
                SELECT mk.sks
                FROM tbnilai n
                JOIN tbkrs k ON n.id_krs = k.id_krs
                JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk
                WHERE n.id_nilai = %s
            """

            cursor.execute(query_sks, (selected_id_nilai,))
            result = cursor.fetchone()
            sks = result[0]
            mutu = sks * bobot

            sql = """
                UPDATE tbnilai
                SET nilai_angka=%s,
                    nilai_huruf=%s,
                    bobot=%s,
                    mutu=%s
                WHERE id_nilai=%s
            """

            data = (
                nilai_angka,
                huruf,
                bobot,
                mutu,
                selected_id_nilai
            )

            cursor.execute(sql, data)
            conn.commit()

            messagebox.showinfo("Sukses", "Data nilai berhasil diupdate.")
            show_data()

        except Exception as e:
            messagebox.showerror("Error", f"Gagal update data: {e}")
        finally:
            if conn: conn.close()


# ==============================
#            TOMBOL
# ==============================
    frame_Button = tk.Frame(frame_input)
    frame_Button.grid(row=2, column=1, columnspan=3, pady=15, sticky="w")

    tk.Button(frame_Button, text="SAVE", command=simpan_nilai, bg="#4CAF50", fg="white", width=10, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="SHOW DATA", command=show_data, bg="#9C27B0", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="DELETE", command=delete_nilai, bg="#F44336", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="UPDATE", command=update_nilai, bg="#FF9800", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="HOME", command=halaman_utama, bg="#9E9E9E", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)

# ==============================
#            TREEVIEW
# ==============================
    frame_table = tk.Frame(root)
    frame_table.pack(pady=10, fill=tk.BOTH, expand=True, padx=20)

    scroll_x = tk.Scrollbar(frame_table, orient=tk.HORIZONTAL)
    scroll_x.pack(side=tk.BOTTOM, fill=tk.X)

    scroll_y = tk.Scrollbar(frame_table, orient=tk.VERTICAL)
    scroll_y.pack(side=tk.RIGHT, fill=tk.Y)

    columns = (
        "id_nilai",
        "nama_mahasiswa",
        "nama_matakuliah",
        "nilai_angka",
        "nilai_huruf",
        "bobot",
        "mutu"
    )

    tree = ttk.Treeview(
        frame_table,
        columns=columns,
        show="headings",
        xscrollcommand=scroll_x.set,
        yscrollcommand=scroll_y.set,
        height=10
    )

    scroll_x.config(command=tree.xview)
    scroll_y.config(command=tree.yview)

    tree.heading("id_nilai", text="No")
    tree.heading("nama_mahasiswa", text="Nama Mahasiswa")
    tree.heading("nama_matakuliah", text="Mata Kuliah")
    tree.heading("nilai_angka", text="Nilai Angka")
    tree.heading("nilai_huruf", text="Nilai Huruf")
    tree.heading("bobot", text="Bobot")
    tree.heading("mutu", text="Mutu")

    tree.column("id_nilai", width=50, anchor="center")
    tree.column("nama_mahasiswa", width=180)
    tree.column("nama_matakuliah", width=180)
    tree.column("nilai_angka", width=100, anchor="center")
    tree.column("nilai_huruf", width=100, anchor="center")
    tree.column("bobot", width=100, anchor="center")
    tree.column("mutu", width=100, anchor="center")

    tree.pack(fill=tk.BOTH, expand=True)

    tree.bind("<ButtonRelease-1>", pilih_data)
    tree.tag_configure("oddrow", background="white")
    tree.tag_configure("evenrow", background="lightblue")
```

```python
# ==========================================
#              FORM LAPORAN KRS 
# ==========================================

def form_laporan_krs():
    clear_frame()

    judul = tk.Label(root, text="LAPORAN KARTU RENCANA STUDI (KRS)", font=("Arial", 16, "bold"), fg="#1A237E")
    judul.pack(pady=20)

    frame_atas = tk.Frame(root)
    frame_atas.pack(pady=10, fill=tk.X, padx=20)

    frame_input = tk.LabelFrame(frame_atas, text=" Filter Laporan KRS ", padx=10, pady=10)
    frame_input.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=(0, 10))

    frame_search = tk.LabelFrame(frame_atas, text=" Pencarian Cepat ", padx=10, pady=10)
    frame_search.pack(side=tk.RIGHT, fill=tk.BOTH, padx=(10, 0))

    def get_list_mahasiswa():
        try:
            conn = get_connection()
            cursor = conn.cursor()
            cursor.execute("SELECT nim, nama_mahasiswa FROM tbmahasiswa")
            return [f"{row[0]} - {row[1]}" for row in cursor.fetchall()]
        except: return []
        finally: conn.close()

    tk.Label(frame_input, text="Pilih Mahasiswa").grid(row=0, column=0, sticky="w", padx=5, pady=5)
    combo_mahasiswa = ttk.Combobox(frame_input, values=get_list_mahasiswa(), width=45, state="readonly")
    combo_mahasiswa.grid(row=0, column=1, padx=5, pady=5, sticky="w")

    tk.Label(frame_search, text="Cari Nama/Mata Kuliah:").pack(anchor="w", pady=2)
    entry_cari = tk.Entry(frame_search, width=25)
    entry_cari.pack(pady=5)

    def bersihkan():
        combo_mahasiswa.set("")
        entry_cari.delete(0, tk.END)
        for item in tree.get_children():
            tree.delete(item)

    def show_data(keyword=None):
        for item in tree.get_children():
            tree.delete(item)
            
        if not combo_mahasiswa.get() and not keyword:
            messagebox.showwarning("Peringatan", "Silahkan pilih mahasiswa terlebih dahulu!")
            return

        try:
            conn = get_connection()
            cursor = conn.cursor()
            
            query = """
                SELECT k.id_krs, k.nim, m.nama_mahasiswa, k.kode_mk, mk.nama_matakuliah, mk.sks,
                       k.kelas, k.ruang, k.hari, k.jam, d.nama_dosen
                FROM tbkrs k
                JOIN tbmahasiswa m ON k.nim = m.nim
                JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk
                JOIN tbdosen d ON k.nuptk = d.nuptk
            """
            
            if combo_mahasiswa.get():
                nim = combo_mahasiswa.get().split(" - ")[0]
                if keyword:
                    query += " WHERE k.nim = %s AND (mk.nama_matakuliah LIKE %s OR d.nama_dosen LIKE %s)"
                    cursor.execute(query, (nim, f"%{keyword}%", f"%{keyword}%"))
                else:
                    query += " WHERE k.nim = %s"
                    cursor.execute(query, (nim,))
            else:
                query += " WHERE m.nama_mahasiswa LIKE %s OR mk.nama_matakuliah LIKE %s"
                cursor.execute(query, (f"%{keyword}%", f"%{keyword}%"))

            rows = cursor.fetchall()
            if not rows:
                messagebox.showinfo("Informasi", "Tidak ada data KRS yang ditemukan.")
                return

            for count, row in enumerate(rows):
                if count % 2 == 0:
                    tree.insert("", "end", values=row, tags=("evenrow",))
                else:
                    tree.insert("", "end", values=row, tags=("oddrow",))
        except Exception as e:
            messagebox.showerror("Error", f"Gagal memuat laporan KRS: {e}")
        finally:
            conn.close()

    def cari_data():
        kata_kunci = entry_cari.get()
        show_data(keyword=kata_kunci)

    def cetak_krs():
        if not combo_mahasiswa.get():
            messagebox.showwarning("Peringatan", "Pilih data mahasiswa yang ingin dicetak!")
            return
        
        items = tree.get_children()
        if not items:
            messagebox.showwarning("Peringatan", "Tampilkan data terlebih dahulu sebelum mencetak!")
            return

        total_sks = 0
        teks_krs = f"KARTU RENCANA STUDI\nMahasiswa: {combo_mahasiswa.get()}\n"
        teks_krs += "="*50 + "\n"
        teks_krs += f"{'Kode MK':<10} | {'Nama Mata Kuliah':<22} | {'SKS':<4}\n"
        teks_krs += "="*50 + "\n"
        
        for item in items:
            val = tree.item(item)["values"]
            teks_krs += f"{val[3]:<10} | {val[4][:20]:<22} | {val[5]:<4}\n"
            total_sks += int(val[5])
            
        teks_krs += "="*50 + "\n"
        teks_krs += f"Total SKS Diambil: {total_sks} SKS"
        
        messagebox.showinfo("Cetak Preview Laporan KRS", teks_krs)

# ==============================
#    FUNGSI EKSPOR KRS KE PDF
# ==============================

    def ekspor_krs_pdf():
        if not combo_mahasiswa.get():
            messagebox.showwarning("Peringatan", "Pilih data mahasiswa yang ingin diekspor!")
            return
        items = tree.get_children()
        if not items:
            messagebox.showwarning("Peringatan", "Tampilkan data terlebih dahulu sebelum mengekspor!")
            return

        from tkinter import filedialog
        from reportlab.lib.pagesizes import letter
        from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
        from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
        from reportlab.lib import colors

        mhs_info = combo_mahasiswa.get().split(" - ")
        nama_file = filedialog.asksaveasfilename(
            defaultextension=".pdf",
            filetypes=[("PDF Files", "*.pdf")],
            initialfile=f"KRS_{mhs_info[0]}.pdf",
            title="Simpan Laporan KRS"
        )
        if not nama_file: return

        doc = SimpleDocTemplate(nama_file, pagesize=letter, rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
        story = []
        styles = getSampleStyleSheet()

        style_cell = ParagraphStyle('CellKRS', parent=styles['Normal'], fontName='Helvetica', fontSize=9, leading=11, alignment=0)
        style_cell_center = ParagraphStyle('CellKRSC', parent=style_cell, alignment=1)
        style_header = ParagraphStyle('HeaderKRS', parent=styles['Normal'], fontName='Helvetica-Bold', fontSize=10, leading=12, textColor=colors.whitesmoke, alignment=1)

        story.append(Paragraph("<b>KARTU RENCANA STUDI (KRS)</b>", styles['Title']))
        story.append(Spacer(1, 15))
        story.append(Paragraph(f"<b>Mahasiswa:</b> {combo_mahasiswa.get()}", styles['Normal']))
        story.append(Spacer(1, 15))

        headers = ["Kode MK", "Nama Mata Kuliah", "SKS", "Kelas", "Hari", "Jam"]
        data_tabel = [[Paragraph(h, style_header) for h in headers]]
        total_sks = 0

        for item in items:
            val = tree.item(item)["values"]
            row = [
                Paragraph(str(val[3]), style_cell_center),
                Paragraph(str(val[4]), style_cell),
                Paragraph(str(val[5]), style_cell_center),
                Paragraph(str(val[6]), style_cell_center),
                Paragraph(str(val[8]), style_cell_center),
                Paragraph(str(val[9]), style_cell_center)
            ]
            data_tabel.append(row)
            total_sks += int(val[5])

        tabel = Table(data_tabel, colWidths=[65, 210, 45, 50, 80, 80], hAlign='CENTER')
        tabel.setStyle(TableStyle([
            ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#1A237E')),
            ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
            ('TOPPADDING', (0,0), (-1,0), 8),
            ('BOTTOMPADDING', (0,0), (-1,0), 8),
            ('ROWBACKGROUNDS', (0,1), (-1,-1), [colors.white, colors.HexColor('#F9F9F9')]),
            ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#D3D3D3')),
            ('TOPPADDING', (0,1), (-1,-1), 6),
            ('BOTTOMPADDING', (0,1), (-1,-1), 6),
        ]))
        story.append(tabel)
        story.append(Spacer(1, 15))
        story.append(Paragraph(f"<b>Total SKS Diambil:</b> {total_sks} SKS", styles['Normal']))

        try:
            doc.build(story)
            messagebox.showinfo("Sukses", "Laporan KRS berhasil diekspor ke PDF!")
        except Exception as e:
            messagebox.showerror("Error", f"Gagal mengekspor PDF: {e}")

    tk.Button(frame_search, text="Search", command=cari_data, bg="lightgray", width=10).pack(pady=5)
    tk.Button(frame_search, text="Refresh", command=lambda: [entry_cari.delete(0, tk.END), show_data()], bg="white", width=10).pack(pady=5)

    frame_Button = tk.Frame(frame_input)
    frame_Button.grid(row=1, column=1, pady=10, sticky="w")

    tk.Button(frame_Button, text="SHOW DATA", command=lambda: show_data(), bg="#9C27B0", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="PRINT KRS", command=cetak_krs, bg="#4CAF50", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="EXPORT PDF", command=ekspor_krs_pdf, bg="#E74C3C", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="CLEAR", command=bersihkan, bg="#FF9800", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="HOME", command=halaman_utama, bg="#9E9E9E", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)

    frame_table = tk.Frame(root)
    frame_table.pack(pady=10, fill=tk.BOTH, expand=True, padx=20)

    scroll_x = tk.Scrollbar(frame_table, orient=tk.HORIZONTAL)
    scroll_x.pack(side=tk.BOTTOM, fill=tk.X)
    scroll_y = tk.Scrollbar(frame_table, orient=tk.VERTICAL)
    scroll_y.pack(side=tk.RIGHT, fill=tk.Y)

    columns = ("id_krs", "nim", "nama_mahasiswa", "kode_mk", "nama_matakuliah", "sks", "kelas", "ruang", "hari", "jam", "nama_dosen")
    tree = ttk.Treeview(frame_table, columns=columns, show="headings", height=8, xscrollcommand=scroll_x.set, yscrollcommand=scroll_y.set)
    
    scroll_x.config(command=tree.xview)
    scroll_y.config(command=tree.yview)

    tree.heading("id_krs", text="ID KRS")
    tree.heading("nim", text="NIM")
    tree.heading("nama_mahasiswa", text="Nama Mahasiswa")
    tree.heading("kode_mk", text="Kode MK")
    tree.heading("nama_matakuliah", text="Nama Mata Kuliah")
    tree.heading("sks", text="SKS")
    tree.heading("kelas", text="Kelas")
    tree.heading("ruang", text="Ruang")
    tree.heading("hari", text="Hari")
    tree.heading("jam", text="Jam")
    tree.heading("nama_dosen", text="Nama Dosen")

    tree.column("id_krs", width=50, anchor="center")
    tree.column("nim", width=80, anchor="center")
    tree.column("nama_mahasiswa", width=140)
    tree.column("kode_mk", width=70, anchor="center")
    tree.column("nama_matakuliah", width=160)
    tree.column("sks", width=40, anchor="center")
    tree.column("kelas", width=50, anchor="center")
    tree.column("ruang", width=60, anchor="center")
    tree.column("hari", width=70, anchor="center")
    tree.column("jam", width=60, anchor="center")
    tree.column("nama_dosen", width=140)

    tree.pack(fill=tk.BOTH, expand=True)
    tree.tag_configure("oddrow", background="white")
    tree.tag_configure("evenrow", background="lightblue")
```

```python
# ==========================================
#              FORM LAPORAN KHS 
# ==========================================

def form_laporan_khs():
    clear_frame()

    judul = tk.Label(root, text="LAPORAN KARTU HASIL STUDI (KHS)", font=("Arial", 16, "bold"), fg="#1A237E")
    judul.pack(pady=20)

    frame_atas = tk.Frame(root)
    frame_atas.pack(pady=10, fill=tk.X, padx=20)

    frame_input = tk.LabelFrame(frame_atas, text=" Filter Laporan KHS ", padx=10, pady=10)
    frame_input.pack(side=tk.LEFT, fill=tk.BOTH, expand=True, padx=(0, 10))

    frame_search = tk.LabelFrame(frame_atas, text=" Info IPS Mahasiswa ", padx=10, pady=10)
    frame_search.pack(side=tk.RIGHT, fill=tk.BOTH, padx=(10, 0))

    def get_list_mahasiswa():
        try:
            conn = get_connection()
            cursor = conn.cursor()
            cursor.execute("SELECT nim, nama_mahasiswa FROM tbmahasiswa")
            return [f"{row[0]} - {row[1]}" for row in cursor.fetchall()]
        except: return []
        finally: conn.close()

    tk.Label(frame_input, text="Pilih Mahasiswa").grid(row=0, column=0, sticky="w", padx=5, pady=5)
    combo_mahasiswa = ttk.Combobox(frame_input, values=get_list_mahasiswa(), width=45, state="readonly")
    combo_mahasiswa.grid(row=0, column=1, padx=5, pady=5, sticky="w")

    lbl_ips_title = tk.Label(frame_search, text="Indeks Prestasi Semester:", font=("Arial", 10, "bold"))
    lbl_ips_title.pack(pady=2)
    lbl_ips_value = tk.Label(frame_search, text="0.00", font=("Arial", 24, "bold"), fg="#4CAF50")
    lbl_ips_value.pack(pady=5)

    def bersihkan():
        combo_mahasiswa.set("")
        lbl_ips_value.config(text="0.00", fg="#4CAF50")
        for item in tree.get_children():
            tree.delete(item)

    def show_data_khs():
        for item in tree.get_children():
            tree.delete(item)
            
        if not combo_mahasiswa.get():
            messagebox.showwarning("Peringatan", "Silahkan pilih mahasiswa terlebih dahulu!")
            return

        nim = combo_mahasiswa.get().split(" - ")[0]

        try:
            conn = get_connection()
            cursor = conn.cursor()
            
            query = """
                SELECT n.id_nilai, mk.kode_mk, mk.nama_matakuliah, mk.sks, 
                       n.nilai_angka, n.nilai_huruf, n.bobot, n.mutu
                FROM tbnilai n
                JOIN tbkrs k ON n.id_krs = k.id_krs
                JOIN tbmatakuliah mk ON k.kode_mk = mk.kode_mk
                WHERE k.nim = %s
            """
            cursor.execute(query, (nim,))
            rows = cursor.fetchall()
            
            if not rows:
                messagebox.showinfo("Informasi", "Nilai untuk mahasiswa ini belum diinput.")
                lbl_ips_value.config(text="0.00", fg="red")
                return

            total_sks = 0
            total_mutu = 0.0

            for count, row in enumerate(rows):
                if count % 2 == 0:
                    tree.insert("", "end", values=row, tags=("evenrow",))
                else:
                    tree.insert("", "end", values=row, tags=("oddrow",))
                
                total_sks += int(row[3])
                total_mutu += float(row[7])

            if total_sks > 0:
                ips = total_mutu / total_sks
                lbl_ips_value.config(text=f"{ips:.2f}", fg="#4CAF50" if ips >= 2.75 else "#FF9800")
            else:
                lbl_ips_value.config(text="0.00")

        except Exception as e:
            messagebox.showerror("Error", f"Gagal memuat data KHS: {e}")
        finally:
            conn.close()

    def cetak_khs():
        if not combo_mahasiswa.get() or not tree.get_children():
            messagebox.showwarning("Peringatan", "Tampilkan data KHS terlebih dahulu!")
            return
            
        teks_khs = f"KARTU HASIL STUDI (KHS)\nMahasiswa: {combo_mahasiswa.get()}\n"
        teks_khs += "="*55 + "\n"
        teks_khs += f"{'Kode':<8} | {'Mata Kuliah':<20} | {'SKS':<4} | {'Huruf':<5} | {'Mutu':<5}\n"
        teks_khs += "="*55 + "\n"
        
        for item in tree.get_children():
            val = tree.item(item)["values"]
            teks_khs += f"{val[1]:<8} | {val[2][:18]:<20} | {val[3]:<4} | {val[5]:<5} | {val[7]:<5}\n"
            
        teks_khs += "="*55 + "\n"
        teks_khs += f"IPS Diperoleh: {lbl_ips_value.cget('text')}"
        
        messagebox.showinfo("Cetak Preview Laporan KHS", teks_khs)

  
    def ekspor_khs_pdf():
        if not combo_mahasiswa.get() or not tree.get_children():
            messagebox.showwarning("Peringatan", "Tampilkan data KHS terlebih dahulu sebelum mengekspor!")
            return

        from tkinter import filedialog
        from reportlab.lib.pagesizes import letter
        from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
        from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
        from reportlab.lib import colors

        mhs_info = combo_mahasiswa.get().split(" - ")
        nama_file = filedialog.asksaveasfilename(
            defaultextension=".pdf",
            filetypes=[("PDF Files", "*.pdf")],
            initialfile=f"KHS_{mhs_info[0]}.pdf",
            title="Simpan Laporan KHS"
        )
        if not nama_file: return

        doc = SimpleDocTemplate(nama_file, pagesize=letter, rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
        story = []
        styles = getSampleStyleSheet()

        style_cell = ParagraphStyle('CellKHS', parent=styles['Normal'], fontName='Helvetica', fontSize=9, leading=11, alignment=0)
        style_cell_center = ParagraphStyle('CellKHSC', parent=style_cell, alignment=1)
        style_header = ParagraphStyle('HeaderKHS', parent=styles['Normal'], fontName='Helvetica-Bold', fontSize=10, leading=12, textColor=colors.whitesmoke, alignment=1)

        story.append(Paragraph("<b>KARTU HASIL STUDI (KHS)</b>", styles['Title']))
        story.append(Spacer(1, 15))
        story.append(Paragraph(f"<b>Mahasiswa:</b> {combo_mahasiswa.get()}", styles['Normal']))
        story.append(Spacer(1, 15))

        headers = ["Kode MK", "Nama Mata Kuliah", "SKS", "Nilai Angka", "Huruf", "Bobot", "Mutu"]
        data_tabel = [[Paragraph(h, style_header) for h in headers]]

        for item in tree.get_children():
            val = tree.item(item)["values"]
            row = [
                Paragraph(str(val[1]), style_cell_center),
                Paragraph(str(val[2]), style_cell),
                Paragraph(str(val[3]), style_cell_center),
                Paragraph(str(val[4]), style_cell_center),
                Paragraph(str(val[5]), style_cell_center),
                Paragraph(str(val[6]), style_cell_center),
                Paragraph(str(val[7]), style_cell_center)
            ]
            data_tabel.append(row)

        tabel = Table(data_tabel, colWidths=[65, 190, 40, 65, 45, 45, 50], hAlign='CENTER')
        tabel.setStyle(TableStyle([
            ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#1A237E')),
            ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
            ('TOPPADDING', (0,0), (-1,0), 8),
            ('BOTTOMPADDING', (0,0), (-1,0), 8),
            ('ROWBACKGROUNDS', (0,1), (-1,-1), [colors.white, colors.HexColor('#F9F9F9')]),
            ('GRID', (0,0), (-1,-1), 0.5, colors.HexColor('#D3D3D3')),
            ('TOPPADDING', (0,1), (-1,-1), 6),
            ('BOTTOMPADDING', (0,1), (-1,-1), 6),
        ]))
        story.append(tabel)
        story.append(Spacer(1, 15))
        story.append(Paragraph(f"<b>Indeks Prestasi Semester (IPS):</b> {lbl_ips_value.cget('text')}", styles['Normal']))

        try:
            doc.build(story)
            messagebox.showinfo("Sukses", "Laporan KHS berhasil diekspor ke PDF!")
        except Exception as e:
            messagebox.showerror("Error", f"Gagal mengekspor PDF: {e}")

    frame_Button = tk.Frame(frame_input)
    frame_Button.grid(row=1, column=1, pady=10, sticky="w")

    tk.Button(frame_Button, text="SHOW DATA", command=show_data_khs, bg="#9C27B0", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="PRINT KHS", command=cetak_khs, bg="#4CAF50", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="EXPORT PDF", command=ekspor_khs_pdf, bg="#E74C3C", fg="white", width=11, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="CLEAR", command=bersihkan, bg="#FF9800", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)
    tk.Button(frame_Button, text="HOME", command=halaman_utama, bg="#9E9E9E", fg="white", width=8, font=("Arial", 9, "bold")).pack(side=tk.LEFT, padx=3)

    frame_table = tk.Frame(root)
    frame_table.pack(pady=10, fill=tk.BOTH, expand=True, padx=20)

    scroll_x = tk.Scrollbar(frame_table, orient=tk.HORIZONTAL)
    scroll_x.pack(side=tk.BOTTOM, fill=tk.X)
    scroll_y = tk.Scrollbar(frame_table, orient=tk.VERTICAL)
    scroll_y.pack(side=tk.RIGHT, fill=tk.Y)

    columns = ("id_nilai", "kode_mk", "nama_matakuliah", "sks", "nilai_angka", "nilai_huruf", "bobot", "mutu")
    tree = ttk.Treeview(frame_table, columns=columns, show="headings", height=10, xscrollcommand=scroll_x.set, yscrollcommand=scroll_y.set)
    
    scroll_x.config(command=tree.xview)
    scroll_y.config(command=tree.yview)

    tree.heading("id_nilai", text="ID Nilai")
    tree.heading("kode_mk", text="Kode MK")
    tree.heading("nama_matakuliah", text="Nama Mata Kuliah")
    tree.heading("sks", text="SKS")
    tree.heading("nilai_angka", text="Nilai Angka")
    tree.heading("nilai_huruf", text="Nilai Huruf")
    tree.heading("bobot", text="Bobot")
    tree.heading("mutu", text="Mutu")

    tree.column("id_nilai", width=60, anchor="center")
    tree.column("kode_mk", width=90, anchor="center")
    tree.column("nama_matakuliah", width=250)
    tree.column("sks", width=60, anchor="center")
    tree.column("nilai_angka", width=90, anchor="center")
    tree.column("nilai_huruf", width=90, anchor="center")
    tree.column("bobot", width=80, anchor="center")
    tree.column("mutu", width=80, anchor="center")

    tree.pack(fill=tk.BOTH, expand=True)
    tree.tag_configure("oddrow", background="white")
    tree.tag_configure("evenrow", background="lightblue")
```
